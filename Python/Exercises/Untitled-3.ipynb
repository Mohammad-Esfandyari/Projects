```python
#Q1
def my_inside_triangle(x,y):
    if x<0 or y<0 or x+y>1 :
        return 'outside'
    elif x==0 or y==0 or x+y==1:
        return 'border'
    else:
        return 'inside'
print(my_inside_triangle(x=int(input()),y=int(input())))
```
