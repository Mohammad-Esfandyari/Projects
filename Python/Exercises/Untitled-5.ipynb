```python
import pandas as pd
```

```python
woodinc = pd.read_excel(r"D:\تحلیل داده\python\تمرین\WoodInc-sale.xlsx")
woodinc
```

```python
group1 = woodinc.groupby('کد مشتري').agg(net = ('مبلغ خالص','sum')).reset_index()
group1.sort_values(by='net',ascending=False)
```

```python
group2 = woodinc.groupby(['کد مشتري','نام مشتري']).agg(net = ('مبلغ خالص','sum')).reset_index()
group2.sort_values(by='net',ascending=False)
group2.iloc[[8,6,338]]
```

```python
x = woodinc['مبلغ خالص'].sum()

group3 = woodinc.groupby(['کد مشتري','نام مشتري']).agg(net = ('مبلغ خالص','sum')).reset_index()
group3['pernet']=group3['net']/x
group3.sort_values(by='pernet',ascending=False)
group3.iloc[[8,6,338],[0,1,3]]
```

```python
group4 = woodinc.groupby('کد کالا').agg(net = ('مبلغ خالص','sum')).reset_index()
group4.sort_values(by='net',ascending=False)
```

```python
group4 = woodinc.groupby(['کد کالا','نام کالا']).agg(net = ('مبلغ خالص','sum')).reset_index()
group4.sort_values(by='net',ascending=False)
group4.iloc[[307,99,419]]
```

```python
y = woodinc['مبلغ خالص'].sum()

group4 = woodinc.groupby(['کد کالا','نام کالا']).agg(net = ('مبلغ خالص','sum')).reset_index()
group4['pernet']=group4['net']/y
group4.sort_values(by='pernet',ascending=False)
group4.iloc[[307,99,419],[0,1,3]]
```
